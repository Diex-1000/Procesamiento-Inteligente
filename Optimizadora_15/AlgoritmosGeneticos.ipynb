# Resumen de la lectura Algoritmos genéticos

La naturaleza es básicamente una optimizadora: a lo largo de millones de años, los organismos que mejor se adaptan a su entorno sobreviven y se reproducen, mientras los que no logran adaptarse desaparecen. Este proceso, la selección natural, mejora las soluciones generación tras generación sin necesidad de un control central. Los algoritmos genéticos toman prestada esta idea y la aplican a la computación: se parte de un conjunto de posibles soluciones, se evalúa cuáles funcionan mejor y se generan nuevas combinando las más prometedoras.

Para entender cómo funcionan, es necesario conocer un poco de biología. Cada solución candidata se llama individuo y su información está codificada en un cromosoma compuesto por genes. Qué tan buena es esa solución se mide con la aptitud (fitness). El proceso sigue un ciclo: se inicializa una población aleatoria, se evalúa cada individuo, se seleccionan los mejores, se combinan con cruzamiento y se les aplican cambios aleatorios llamados mutaciones. Esto se repite hasta llegar a una solución aceptable.

```python
import random

class Individuo:
    def __init__(self, longitud_cromosoma: int):
        self.cromosoma = [random.randint(0, 1) for _ in range(longitud_cromosoma)]
        self.aptitud = 0.0

    def calcular_aptitud(self):
        self.aptitud = sum(self.cromosoma)
        return self.aptitud

    def mutar(self, tasa_mutacion: float = 0.01):
        for i in range(len(self.cromosoma)):
            if random.random() < tasa_mutacion:
                self.cromosoma[i] = 1 - self.cromosoma[i]

    def mostrar(self):
        print(f"Cromosoma: {self.cromosoma}  |  Aptitud: {self.aptitud}")


class Poblacion:
    def __init__(self, tamanio: int, longitud_cromosoma: int):
        self.tamanio = tamanio
        self.longitud_cromosoma = longitud_cromosoma
        self.generacion = 0
        self.individuos = []

    def inicializar(self):
        self.individuos = [Individuo(self.longitud_cromosoma) for _ in range(self.tamanio)]
        for ind in self.individuos:
            ind.calcular_aptitud()

    def seleccionar(self):
        candidatos = random.sample(self.individuos, 2)
        return max(candidatos, key=lambda ind: ind.aptitud)

    def cruzar(self, padre: Individuo, madre: Individuo) -> Individuo:
        punto = random.randint(1, self.longitud_cromosoma - 1)
        hijo = Individuo(self.longitud_cromosoma)
        hijo.cromosoma = padre.cromosoma[:punto] + madre.cromosoma[punto:]
        return hijo

    def evolucionar(self, tasa_mutacion: float = 0.01):
        nueva_poblacion = []
        for _ in range(self.tamanio):
            padre = self.seleccionar()
            madre = self.seleccionar()
            hijo = self.cruzar(padre, madre)
            hijo.mutar(tasa_mutacion)
            hijo.calcular_aptitud()
            nueva_poblacion.append(hijo)
        self.individuos = nueva_poblacion
        self.generacion += 1

    def mejor_individuo(self) -> Individuo:
        return max(self.individuos, key=lambda ind: ind.aptitud)

    def mostrar_estadisticas(self):
        mejor = self.mejor_individuo()
        promedio = sum(i.aptitud for i in self.individuos) / self.tamanio
        print(f"Generación {self.generacion} | Mejor aptitud: {mejor.aptitud} | Promedio: {promedio:.2f}")
```
